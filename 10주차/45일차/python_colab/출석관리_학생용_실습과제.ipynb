## 1. 실습 개요

**상황:** 오늘 수업의 출석 기록을 프로그램으로 관리하려고 합니다. 학생 이름이 여러 번 입력될 수 있으므로 **기록 수**와 **실제 출석 학생 수**를 구분해야 합니다.

**초기 입력 데이터**

> 💡 **순서의 이유:** 리스트와 튜플은 '순서 있는 묶음'이라는 점이 같고 **바꿀 수 있느냐**만 다릅니다. 집합은 **중복과 순서**를 버리고, 딕셔너리는 **이름표(키)** 를 붙입니다. 비슷한 것끼리 붙여서 비교하며 배웁니다.

## 2. 준비 — 자동 점검 도구 실행

````python
# ▶ 이 셀을 가장 먼저 실행하세요. (자동 점검 도구를 준비합니다)
def check(no, desc, condition, hint=""):
    """condition()이 참이면 ✅, 아니면 ❌와 힌트를 보여 줍니다."""
    try:
        ok = bool(condition())
    except Exception as e:
        ok = False
        hint = f"{hint} / 오류: {type(e).__name__} → 위 셀을 먼저 실행했는지, 빈칸을 채웠는지 확인하세요"
    if ok:
        print(f"✅ [{no}] {desc}")
    else:
        print(f"❌ [{no}] {desc}  → 힌트: {hint}")

print("준비 완료! 이제 아래 셀을 차례대로 실행하세요.")
````

## 7. 실습 5 — 같은 데이터, 네 가지 자료형 *(비교 실험)*
똑같은 출석 기록 `["김민수", "이영희", "박철수", "김민수"]`를 네 가지 자료형에 넣고, **같은 동작**을 시켜 봅니다.

````python
# ▶ 실행만 하세요. 코드를 모두 이해하지 않아도 됩니다. '결과 표'를 관찰하세요.
import copy, operator
import pandas as pd

data = ["김민수", "이영희", "박철수", "김민수"]      # 똑같은 출석 기록 4건

samples = {
    "리스트 list":   list(data),
    "튜플 tuple":    tuple(data),
    "집합 set":      set(data),
    "딕셔너리 dict": dict.fromkeys(data, "출석"),     # 이름(키) → '출석'(값)
}

# 자료형별로 '추가'와 '변경'을 시도하는 방법
add_try = {
    list:  lambda x: x.append("최유진"),
    tuple: lambda x: x.append("최유진"),
    set:   lambda x: x.add("최유진"),
    dict:  lambda x: x.update({"최유진": "출석"}),
}
change_try = {
    list:  lambda x: operator.setitem(x, 0, "최유진"),        # x[0] = "최유진"
    tuple: lambda x: operator.setitem(x, 0, "최유진"),        # x[0] = "최유진"
    set:   lambda x: operator.setitem(x, 0, "최유진"),        # x[0] = "최유진"
    dict:  lambda x: operator.setitem(x, "김민수", "지각"),    # x["김민수"] = "지각"
}

def can(action, obj):
    """복사본으로 시도해 보고, 되면 ⭕ / 오류면 ❌(오류 이름)"""
    try:
        action(copy.copy(obj))
        return "⭕"
    except Exception as e:
        return f"❌ {type(e).__name__}"

rows = []
for name, obj in samples.items():
    rows.append({
        "자료형": name,
        "내용": str(obj),
        "len()": len(obj),
        "중복 보존": "⭕" if len(obj) == len(data) else "❌ (하나로 합침)",
        "'김민수' in": "김민수" in obj,
        "번호 [0]로 꺼내기": "⭕ (키로 꺼냄)" if isinstance(obj, dict) else can(lambda x: x[0], obj),
        "항목 추가": can(add_try[type(obj)], obj),
        "항목 변경": can(change_try[type(obj)], obj),
    })

pd.DataFrame(rows).set_index("자료형")
````

## 8. 실습 6 — 🚀 도전: 파이썬의 다른 '비슷한' 자료형

### 🔎 파이썬에는 '비슷한' 자료형이 더 있습니다

네 가지 기본 자료형을 **조금씩 변형한 친척들**입니다. 이름 옆 괄호는 어디에 있는지(내장 / `collections` 모듈)를 뜻합니다.

| 자료형 | 닮은 기본 자료형 | 무엇이 다른가 | 출석 관리에서의 예 |
|---|---|---|---|
| **frozenset** (내장) | 집합 | **바꿀 수 없는** 집합 (리스트 ↔ 튜플 관계와 같음) | 학기 초에 확정된 수강생 명단 |
| **Counter** (collections) | 딕셔너리 | **개수를 세는** 딕셔너리 | 학생별 출석 기록 횟수 |
| **namedtuple** (collections) | 튜플 | 항목에 **이름표**가 붙은 튜플 | (이름, 번호, 상태) 학생 카드 |
| defaultdict (collections) | 딕셔너리 | 없는 키를 조회하면 **기본값을 자동 생성** | 날짜별 출석자 목록 |
| deque (collections) | 리스트 | **양쪽 끝**에서 넣고 빼기가 빠름 | 상담 대기 줄 |
| str, range (내장) | 튜플 | 순서 있고 바꿀 수 없는 '글자·숫자' 묶음 | 학번 문자열, 번호 1~30 |

이번 도전에서는 굵게 표시한 **Counter, frozenset, namedtuple** 세 가지를 직접 실행해 봅니다.

### 6-1. Counter — 개수를 세는 딕셔너리
리스트(기록 건수) + 집합(학생 수) + 딕셔너리(이름 → 횟수)를 **한 번에** 해 줍니다.

````python
from collections import Counter

sample_log = ["김민수", "이영희", "박철수", "김민수", "최유진", "이영희", "김민수"]
counts = Counter(sample_log)           # 이름 → 기록 횟수 를 자동으로 셈

print(counts)
print("김민수 기록 횟수:", counts["김민수"])
print("전체 기록 건수:", sum(counts.values()))   # 리스트의 len()과 같은 값
print("서로 다른 학생 수:", len(counts))          # 집합의 len()과 같은 값
````

### 6-2. frozenset — 바꿀 수 없는 집합
**리스트를 바꿀 수 없게 만든 것이 튜플**이듯, **집합을 바꿀 수 없게 만든 것이 frozenset**입니다.

````python
semester_roster = frozenset(["김민수", "이영희", "박철수", "최유진"])   # 학기 초 확정 명단

print(semester_roster)
print("이영희는 수강생인가?", "이영희" in semester_roster)

try:
    semester_roster.add("한지우")                 # 바꿀 수 없으므로 오류
except AttributeError as error:
    print("❗ 오류 발생:", error)

today = {"김민수", "박철수"}
print("결석자(집합 연산은 그대로 가능):", semester_roster - today)
````

### 6-3. namedtuple — 이름표가 붙은 튜플
`s1[2]` 보다 `s1.status` 가 읽기 쉽습니다.

````python
from collections import namedtuple

Student = namedtuple("Student", ["name", "number", "status"])   # 이름표: 이름, 번호, 상태

s1 = Student("김민수", 1, "출석")
print(s1)
print("이름표로 꺼내기:", s1.name, s1.status)
print("번호로 꺼내기(튜플처럼):", s1[0], s1[2])
````

### 6-4. (참고) defaultdict, deque

````python
# (참고 — 실행만) defaultdict 와 deque 맛보기
from collections import defaultdict, deque

by_date = defaultdict(list)                  # 없는 날짜를 조회하면 빈 리스트를 자동으로 만듦
by_date["10/20"].append("김민수")
by_date["10/20"].append("이영희")
by_date["10/21"].append("박철수")
print("날짜별 출석자:", dict(by_date))

waiting = deque(["김민수", "이영희"])         # 상담 대기 줄
waiting.append("박철수")                      # 뒤에 줄 서기
print("먼저 상담할 학생:", waiting.popleft())  # 앞에서 빼기
print("남은 대기 줄:", list(waiting))
````
