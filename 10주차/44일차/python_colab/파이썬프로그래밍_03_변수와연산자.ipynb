## 2. 산술 연산자
| 연산자 | 의미 | 예 (a=10, b=3) |
|---|---|---|
| `+` `-` `*` | 더하기, 빼기, 곱하기 | 13, 7, 30 |
| `/` | 나누기 (실수 결과) | 3.333… |
| `//` | 몫 | 3 |
| `%` | 나머지 | 1 |
| `**` | 거듭제곱 | 1000 |

```python
a, b = 10, 3
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)
print("a // b =", a // b)
print("a % b  =", a % b)
print("a ** b =", a ** b)
```

### 실습: 편의점 계산기
콜라(1,500원) 4개와 햄버거(3,000원) 2개를 사고 10% 할인을 받았을 때 결제 금액은?

```python
cola, hamburger = 1500, 3000
nCola, nHamburger = 4, 2
discount = 0.1

total = (cola * nCola + hamburger * nHamburger) * (1 - discount)
print("결제 금액:", total, "원")
```

### 실습: 2+1 행사
콜라를 3개 단위로 사면 1개를 덤으로 줍니다. `//`로 덤 개수를, `%`로 다음 행사까지 더 사야 할 개수를 구합니다.

```python
nCola = 7
nEvent = nCola // 3                 # 덤으로 받는 개수
N_lack_cola = 3 - (nCola % 3)       # 다음 덤까지 더 사야 할 개수

print("덤으로 받는 콜라:", nEvent, "개")
print("한 개 더 받으려면", N_lack_cola, "개를 더 사야 합니다.")
```

## 3. 관계(비교) 연산자
두 값을 비교하여 `True` 또는 `False`를 돌려줍니다.

`==` 같다, `!=` 다르다, `>` `<` `>=` `<=` 크기 비교

```python
age = 19
print("성인인가?", 20 <= age)
print("19세인가?", age == 19)
print("20세가 아닌가?", age != 20)
```

```python
year_of_birth = 1997
print("구매 가능?", year_of_birth % 10 == 7)
```

## 4. 논리 연산자
| A | B | A and B | A or B | not A |
|---|---|---|---|---|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

```python
A, B = True, False
print("A and B:", A and B)
print("A or B :", A or B)
print("not A  :", not A)
```

## 5. 복합 대입 연산자
`a = a + 3`을 `a += 3`처럼 줄여 쓸 수 있습니다. (`-=`, `*=`, `/=`, `//=`, `%=`, `**=`도 같음)

```python
a = 10
a += 3;  print("a += 3  →", a)
a -= 1;  print("a -= 1  →", a)
a *= 2;  print("a *= 2  →", a)
a //= 5; print("a //= 5 →", a)
```

### 실습: 바코드 스캔 누적 합계
계산대에서 물건을 하나씩 찍을 때마다 합계가 늘어나는 모습을 복합 대입 연산자로 표현합니다.

```python
total = 0
total += 1500   # 콜라
total += 3000   # 햄버거
total += 1200   # 과자
print("합계:", total, "원")
```

## 6. 연산자 우선순위
높은 순서: `**` → 부호(`-x`) → `*` `/` `//` `%` → `+` `-` → 비교 연산자 → `not` → `and` → `or`

헷갈릴 때는 **괄호 `( )`** 를 사용하면 가장 먼저 계산됩니다.

```python
print(2 + 3 * 4)        # 곱셈 먼저
print((2 + 3) * 4)      # 괄호 먼저
print(-2 ** 2)          # ** 가 부호보다 먼저 → -(2**2)
print(3 > 2 and 1 > 2 or True)
```
