## 1. 튜플(tuple)
리스트와 비슷하지만 **값을 바꿀 수 없는** 자료형입니다.

| | 리스트 | 튜플 |
|---|---|---|
| 기호 | `[ ]` | `( )` |
| 추가·삭제·수정 | 가능 | **불가능** |

- 요소가 1개일 때는 `(1,)`처럼 **쉼표를 꼭** 붙입니다.
- 괄호를 생략해도 튜플이 됩니다. `t4 = 1, 2, 3`
- 프로그램이 도는 동안 바뀌면 안 되는 값은 튜플, 수시로 바뀌는 값은 리스트를 사용합니다.

```python
t1 = ()
t2 = (1,)
t3 = (1, 2, 3)
t4 = 1, 2, 3
t5 = ("a", "b", ("ab", "cd"))
print(t1, t2, t3, t4, t5)
print(type((1)), type((1,)))     # 쉼표가 없으면 그냥 정수!
```

### 튜플의 값은 지우거나 바꿀 수 없다
아래 코드는 일부러 오류를 일으킵니다. (`try/except`는 오류가 나도 멈추지 않고 메시지만 보여 주기 위해 사용했습니다.)

```python
t1 = (1, 2, "a", "b")
try:
    del t1[0]
except TypeError as e:
    print("삭제 오류:", e)

try:
    t1[0] = "c"
except TypeError as e:
    print("수정 오류:", e)
```

### 튜플 다루기: 인덱싱, 슬라이싱, 더하기, 곱하기, 길이

```python
t1 = (1, 2, "a", "b")
t2 = (3, 4)
print(t1[0], t1[3])
print(t1[1:])
print(t1 + t2)
print(t2 * 3)
print(len(t1))
```

### 실습: 온라인 도서 주문 명세서
한 번 확정된 주문 내역은 바뀌면 안 되므로 튜플로 저장합니다.

```python
transaction1001 = ("파이썬 코딩 길잡이", "데이터베이스론", "JAVA")
transaction1002 = ("요리비법", "영화평론", "해리포터")
print("3월 1일 주문 송장 :", transaction1001)
print("3월 15일 주문 송장:", transaction1002)
```

## 2. 딕셔너리(dictionary)
**Key(키)와 Value(값)** 를 한 쌍으로 저장하는 자료형입니다. 예) 햄버거 → 3,000
순서로 찾지 않고 **키로 값을 바로** 찾습니다. 형태: `{Key1: Value1, Key2: Value2}`

```python
dic = {"name": "pey", "age": 25, "birth": "0328"}
a = {1: "hi"}
b = {"a": [1, 2, 3]}
print(dic, a, b, sep="\n")
print(dic["name"])
```

### 쌍 추가·삭제, 키로 값 얻기

```python
a = {1: "a"}
a[2] = "b"
a["name"] = "pey"
print("추가:", a)
del a[1]
print("삭제:", a)

score = {"pey": 40, "juliet": 99, "tom": 70}
print("pey 점수:", score["pey"], "/ tom 점수:", score["tom"])
```

### 딕셔너리 관련 함수
| 함수 | 기능 |
|---|---|
| `keys()` | 키 목록 |
| `values()` | 값 목록 |
| `items()` | (키, 값) 쌍 목록 |
| `get(key)` | 값 반환 (키가 없으면 `None`, 오류 없음) |
| `in` | 키가 있는지 확인 |
| `clear()` | 모두 삭제 |

```python
a = {"name": "pey", "phone": "01022223333", "birth": "1118"}
print(a.keys())
print(list(a.keys()))
print(a.values())
print(a.items())
print(a.get("name"), a.get("nokey"))
print(a.get("email", "정보 없음"))      # 기본값 지정
print("name" in a, "email" in a, "email" not in a)
```

```python
for key, value in a.items():
    print(f"{key:>6} : {value}")
a.clear()
print("clear 후:", a)
```

## 3. 집합(set)
`set()`으로 만들며 두 가지 특징이 있습니다.
1. **중복을 허용하지 않는다** → 중복 제거에 유용
2. **순서가 없다** → 인덱싱 불가 (필요하면 `list()`나 `tuple()`로 변환)

```python
s1 = set([1, 2, 3])
s2 = set("Hello")
print(s1)
print(s2)               # 'l'이 한 번만 남음

l1 = list(s1)
t1 = tuple(s1)
print(l1[0], t1[0])

print(set([1, 1, 2, 2, 3]))     # 리스트의 중복 제거
```

### 교집합·합집합·차집합
| 연산 | 기호 | 함수 |
|---|---|---|
| 교집합 | `&` | `intersection()` |
| 합집합 | `\|` | `union()` |
| 차집합 | `-` | `difference()` |

```python
s1 = set([1, 2, 3, 4, 5, 6])
s2 = set([4, 5, 6, 7, 8, 9])
print("교집합:", s1 & s2, s1.intersection(s2))
print("합집합:", s1 | s2, s1.union(s2))
print("차집합:", s1 - s2, s1.difference(s2))
```

### 집합 관련 함수: `add()` 1개 추가, `update()` 여러 개 추가, `remove()` 제거

```python
s1 = set([1, 2, 3])
s1.add(4);             print(s1)
s1.update([5, 6, 7]);  print(s1)
s1.remove(2);          print(s1)
```

## 4. 불(bool)
참(`True`)과 거짓(`False`) 두 값만 가지는 자료형입니다. 비교 연산의 결과로도 나옵니다.

```python
a = True
b = False
print(type(a), type(b))
print(1 == 1, 2 > 1, 2 < 1)
```

### 자료형의 참과 거짓
**비어 있거나 0이면 거짓**, 그 외에는 참입니다. `bool()`로 확인합니다.

| 값 | 참/거짓 |
|---|---|
| `"python"` | 참 |
| `""` | 거짓 |
| `[1, 2, 3]` | 참 |
| `[]`, `()`, `{}` | 거짓 |
| `1`, `3` | 참 |
| `0` | 거짓 |
| `None` | 거짓 |

```python
for v in ["python", "", [1, 2, 3], [], (), {}, 0, 3, None]:
    print(f"bool({v!r:>11}) = {bool(v)}")
```

```python
# 활용: 리스트가 빌 때까지 꺼내기
a = [1, 2, 3, 4]
while a:
    print("꺼낸 값:", a.pop())
print("리스트가 비었습니다.")
```

## 5. 자료형 종합 실습
학생 정보를 **딕셔너리**로, 여러 학생을 **리스트**로, 수강 과목을 **집합**으로 관리해 봅시다.

```python
students = [
    {"name": "김철수", "scores": (90, 80, 70), "subjects": {"파이썬", "수학"}},
    {"name": "이영희", "scores": (85, 95, 100), "subjects": {"파이썬", "영어"}},
]

for s in students:
    avg = sum(s["scores"]) / len(s["scores"])
    print(f"{s['name']} 평균: {avg:.1f}")

common = students[0]["subjects"] & students[1]["subjects"]
print("두 학생이 함께 듣는 과목:", common)
```
