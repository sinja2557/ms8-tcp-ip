## 1. 상속이란?
기존 클래스(**부모**)의 속성과 메서드를 물려받아 새 클래스(**자식**)를 만들고, 필요한 부분만 고치거나 추가하는 것입니다.

예) 부모 `Country` → 자식 `Korea`, `USA`

```python
class 자식클래스(부모클래스):
    ...
```

## 2. 생성자 `__init__`
객체가 만들어질 때 **자동으로 한 번 실행**되는 메서드입니다.

````python
class Hello:
    def __init__(self):
        print("생성자가 수행되었습니다.")

a = Hello()
````

## 3. 부모 클래스 정의

````python
class Country:
    name = "국가명"
    population = "인구"
    capital = "수도"

    def show(self):
        print("국가 클래스의 메서드입니다.")
````

## 4. 자식 클래스 정의
`Korea`는 `Country`를 상속받아 `show()`, `capital`, `population`을 그대로 쓰고,
`__init__`과 `show_name()`을 새로 추가합니다.

````python
class Korea(Country):
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print("국가 이름은 :", self.name)

a = Korea("대한민국")
a.show()               # 부모의 메서드
a.show_name()          # 자식의 메서드
print(a.capital)       # 부모의 클래스 변수
print(a.population)
print(a.name)          # 자식에서 지정한 값
````

````python
print(isinstance(a, Korea), isinstance(a, Country))
print(issubclass(Korea, Country))
````

## 5. 메서드 오버라이딩
부모의 메서드와 **같은 이름**의 메서드를 자식에서 다시 정의하면, 자식의 것이 사용됩니다.

````python
class Korea(Country):
    def __init__(self, name, population, capital):
        self.name = name
        self.population = population
        self.capital = capital

    def show(self):                       # 부모의 show를 덮어씀
        print(f"""
국가의 이름은 {self.name} 입니다.
국가의 인구는 {self.population:,} 명입니다.
국가의 수도는 {self.capital} 입니다.""")

a = Korea("대한민국", 50000000, "서울")
a.show()
````

## 6. super() – 부모의 메서드 호출
오버라이딩한 메서드 안에서도 `super().메서드()`로 부모의 원래 기능을 함께 실행할 수 있습니다.

````python
class Korea(Country):
    def __init__(self, name, population, capital):
        self.name = name
        self.population = population
        self.capital = capital

    def show(self):
        super().show()                    # 부모 메서드 먼저 실행
        print(f"{self.name}의 수도는 {self.capital}입니다.")

Korea("대한민국", 50000000, "서울").show()
````

## 7. 실습: 상속으로 기능 확장하기
10장의 `FourCal`을 상속받아 거듭제곱(`pow`)을 추가하고, 0으로 나누면 0을 돌려주도록 `div`를 오버라이딩합니다.

````python
class FourCal:
    def __init__(self, first, second):
        self.first = first
        self.second = second
    def add(self): return self.first + self.second
    def div(self): return self.first / self.second

class MoreFourCal(FourCal):
    def pow(self):
        return self.first ** self.second

class SafeFourCal(MoreFourCal):
    def div(self):
        if self.second == 0:
            return 0
        return super().div()

a = MoreFourCal(4, 2)
print(a.add(), a.div(), a.pow())

b = SafeFourCal(4, 0)
print(b.div(), b.pow())
````

### 실습: 여러 자식 클래스
같은 부모를 상속받은 클래스들을 리스트에 담아 한 번에 처리할 수 있습니다.

````python
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "..."
    def introduce(self):
        print(f"{self.name}: {self.speak()}")

class Dog(Animal):
    def speak(self): return "멍멍"

class Cat(Animal):
    def speak(self): return "야옹"

for pet in [Dog("바둑이"), Cat("나비"), Animal("이름 모를 동물")]:
    pet.introduce()
````
