## 5. PCM (표본화 → 양자화 → 부호화)

```
 아날로그 m(t) ──▶ [① 표본화 fs] ──▶ [② 양자화 n비트] ──▶ [③ 부호화] ──▶ 0110 1001 0111 …
```

### 5-1. PCM 설계 📝

| 단계 | 항목 | 기호 | 내가 선택한 값 | 단위 | 이 값을 선택한 이유 |
|---|---|---|---|---|---|
| 표본화 | 표본화 주파수 | fs | (작성) | Hz | (작성) ※ 나이퀴스트 조건 fs ≥ 2fm |
| 양자화 | 양자화 비트 수 | n | (작성) | bit | (작성) |
| 양자화 | 양자화 방식 | - | ☐ 균일(선형) ☐ 비균일(μ-law 등) | - | (작성) |
| 양자화 | 진폭 범위 | - | (작성) ~ (작성) | - | (작성) |
| 부호화 | 부호화 방식 | - | ☐ 자연 2진 ☐ 부호-크기 ☐ 2의 보수 ☐ 기타 | - | (작성) |
| 출력 | 일련 데이터 표시 범위 | - | (작성) 번째 표본부터 (작성) 개 | - | (작성) |

**설계 단계 예상값 (계산)**

| 항목 | 계산식 | 예상값 | 단위 |
|---|---|---|---|
| 표본화 주기 | Ts = 1 / fs | (작성) | μs |
| 나이퀴스트 율 | 2fm | (작성) | Hz |
| 양자화 레벨 수 | L = 2ⁿ | (작성) | 개 |
| 양자화 간격 | Δ = (진폭 범위) / L | (작성) | - |
| 이론 SQNR | 6.02n + 1.76 | (작성) | dB |
| 비트율 | R = fs × n | (작성) | bps |
