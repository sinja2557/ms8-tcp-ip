## 1. 조건문이 필요한 이유
실시간으로 들어오는 혈압 데이터 `{'age': 25, 'sex': 'M', 'maxBP': 130, 'minBP': 90}`를 보고
"정상인지, 정밀 검사가 필요한지" 판단하려면 **조건에 따라 다른 일을 하는** 코드가 필요합니다.
이것이 조건문(`if`)입니다.

## 2. if 문의 기본 구조
```python
if 조건:
    조건이 참일 때 수행할 문장들
else:
    조건이 거짓일 때 수행할 문장들
```
- 조건 뒤에 **콜론(`:`)** 을 붙입니다.
- 블록은 **들여쓰기(공백 4칸)** 로 구분합니다.

````python
x, y = 3, 2
print(x > y)
if x > y:
    print("if 문 수행!")
````

### if – else
"3,000원 이상 있으면 택시를 타고, 그렇지 않으면 걸어가라"

````python
money = 2000
if money >= 3000:
    print("택시를 타고 가라")
else:
    print("걸어 가라")
````

### 조건에 `and`, `or`, `in`을 함께 쓰기
`x in 리스트`는 리스트 안에 x가 있으면 `True`입니다.

````python
money = 2000
card = True
if money >= 3000 or card:
    print("택시를 타고 가라")
else:
    print("걸어 가라")

pocket = ["paper", "cellphone", "money"]
if "money" in pocket:
    print("주머니에 돈이 있다")
````

## 3. elif – 여러 조건을 차례로 판단
"주머니에 돈이 있으면 택시, 돈은 없지만 카드가 있으면 택시, 둘 다 없으면 걸어가라"

`if`와 `else`만으로 작성하면 `if` 안에 `if`가 들어가 복잡해집니다.
`elif`는 **앞의 조건이 거짓일 때만** 검사되며, 개수 제한 없이 쓸 수 있습니다.

````python
# elif 없이 작성 (중첩 if)
pocket = ["paper", "cellphone"]
card = True
if "money" in pocket:
    print("택시를 타고 가라")
else:
    if card:
        print("택시를 타고 가라")
    else:
        print("걸어 가라")
````

````python
# elif 사용 → 훨씬 읽기 쉽다
pocket = ["paper", "cellphone"]
card = True
if "money" in pocket:
    print("택시를 타고 가라")
elif card:
    print("택시를 타고 가라")
else:
    print("걸어 가라")
````

### 실습: 점수에 따른 학점

````python
score = 83
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
print(f"{score}점 → {grade}학점")
````

### 조건부 표현식 (한 줄 if)
`참일 때 값 if 조건 else 거짓일 때 값`

````python
score = 55
result = "합격" if score >= 60 else "불합격"
print(result)
````

## 4. 실습: 혈압 데이터의 정상 여부 판별
20대 이상 기준 정상 혈압(가정)

| 성별 | 수축기(최고) | 이완기(최저) |
|---|---|---|
| 남성 | 120 미만 | 80 미만 |
| 여성 | 115 미만 | 75 미만 |

측정값을 딕셔너리로 받아 **중첩 조건문**으로 판별합니다.
성별이 `M`/`F`가 아니면 잘못 입력했다고 알려 줍니다(예외 처리).

````python
bloodPressure = {"age": 25, "sex": "M", "maxBP": 130, "minBP": 90}

if bloodPressure["age"] >= 20:
    if bloodPressure["sex"] == "M":
        if bloodPressure["maxBP"] < 120 and bloodPressure["minBP"] < 80:
            print("정상 혈압입니다.")
        else:
            print("비정상 혈압입니다. 정밀 검사가 필요합니다.")
    elif bloodPressure["sex"] == "F":
        if bloodPressure["maxBP"] < 115 and bloodPressure["minBP"] < 75:
            print("정상 혈압입니다.")
        else:
            print("비정상 혈압입니다. 정밀 검사가 필요합니다.")
    else:
        print("성별을 잘못 입력하였습니다.")
else:
    print("20세 미만은 별도 기준으로 판단합니다.")
````
