## 0. 준비 — 자동 점검 도구 실행

```python
# ▶ 이 셀을 가장 먼저 실행하세요. (자동 점검 도구를 준비합니다)
def check(no, desc, condition, hint=""):
    """condition()이 참이면 ✅, 아니면 ❌와 힌트를 보여 줍니다."""
    try:
        ok = bool(condition())
    except Exception as e:
        ok = False
        hint = f"{hint} / 오류: {type(e).__name__} → 위 셀을 먼저 실행했는지, 빈칸을 채웠는지 확인하세요"
    if ok:
        print(f"✅ [{no}] {desc}")
    else:
        print(f"❌ [{no}] {desc}  → 힌트: {hint}")

print("준비 완료! 이제 아래 셀을 차례대로 실행하세요.")
```

## 1. 네 가지 자료형의 기본 속성 비교

| 항목 | 리스트 (list) | 튜플 (tuple) | 집합 (set) | 딕셔너리 (dict) |
|---|---|---|---|---|
| 표기 예 | `[10, 20, 10]` | `(10, 20, 10)` | `{10, 20}` | `{'a': 10, 'b': 20}` |
| 저장 형태 | 값의 나열 | 값의 나열 | 중복 없는 값 | 키: 값 쌍 |
| 순서 | 있음 | 있음 | **없음** | 삽입 순서 유지 \* |
| 중복 | 허용 | 허용 | **불허** | 키는 불가 \*\* |
| 변경 | 가능 | **불가능** | 가능(추가·삭제) | 가능 |
| 인덱스 접근 | `x[0]` | `x[0]` | **불가능** | 키로: `x['a']` |
| 빈 자료형 | `[]` | `()` | `set()` | `{}` |
| 주요 용도 | 순서 있는 목록 | 고정된 묶음 | 중복 제거·집합 연산 | 이름표가 있는 정보 |

\* 딕셔너리는 Python 3.7 이상에서 삽입 순서를 보장합니다.
\*\* 값은 중복될 수 있지만 키는 중복될 수 없습니다. 같은 키를 다시 지정하면 기존 값이 바뀝니다.

> 💡 **비교 순서:** 리스트와 튜플은 '바꿀 수 있느냐'만 다르고, 집합은 중복과 순서를 버리며, 딕셔너리는 이름표(키)를 붙입니다. 비슷한 것끼리 나란히 놓았습니다.
>
> ⚠️ `{}`는 빈 **딕셔너리**입니다. 빈 집합은 반드시 `set()`.

## 2. 공통 기능을 문법으로 비교하기

같은 목적의 기능을 자료형마다 어떻게 쓰는지 비교합니다. 아래 셀을 **실행**하면 각 문법을 실제로 실행한 결과가 표로 나옵니다.
- `→` 뒤는 실행 결과(또는 실행 후의 `x`)입니다.
- `❌`는 그 자료형에서 쓸 수 없는 문법이라 **오류**가 난 것입니다.

```python
# ▶ 실행만 하세요. 각 칸의 코드를 실제로 실행한 결과입니다.
import copy
import pandas as pd
pd.set_option("display.max_colwidth", None)

start = {"리스트": [1, 2, 3], "튜플": (1, 2, 3), "집합": {1, 2, 3}, "딕셔너리": {"a": 1, "b": 2}}

table = [
    ("만들기",          ["x", "x", "x", "x"]),
    ("항목 수",         ["len(x)", "len(x)", "len(x)", "len(x)"]),
    ("포함 여부",       ["2 in x", "2 in x", "2 in x", "'a' in x"]),
    ("첫 항목",         ["x[0]", "x[0]", "x[0]", "x['a']"]),
    ("특정 값 찾기",    ["x.index(2)", "x.index(2)", "x.index(2)", "[k for k, v in x.items() if v == 2]"]),
    ("항목 추가",       ["x.append(4)", "x.append(4)", "x.add(4)", "x['c'] = 3"]),
    ("여러 항목 추가",  ["x.extend([4, 5])", "x + (4, 5)", "x.update([4, 5])", "x.update({'c': 3})"]),
    ("항목 삭제",       ["x.remove(2)", "x.remove(2)", "x.discard(2)", "del x['a']"]),
    ("없는 값 삭제",    ["x.remove(9)", "x.remove(9)", "x.discard(9)", "x.pop('z', None)"]),
    ("전체 비우기",     ["x.clear()", "x.clear()", "x.clear()", "x.clear()"]),
    ("항목 반복",       ["[v for v in x]", "[v for v in x]", "[v for v in x]", "[k for k in x]"]),
    ("키·값 함께 반복", ["-", "-", "-", "[(k, v) for k, v in x.items()]"]),
    ("복사",            ["x.copy()", "tuple(x)", "x.copy()", "x.copy()"]),
]

def run(src, x):
    if src == "-":
        return "-"
    env = {"x": copy.copy(x)}
    try:
        try:
            result = eval(src, env)
            out = env["x"] if result is None else result
        except SyntaxError:
            exec(src, env)                      # 대입문·del 은 exec 로 실행
            out = env["x"]
        return f"{src}  →  {out!r}"
    except Exception as e:
        return f"{src}  →  ❌ {type(e).__name__}"

rows = {name: {t: run(src, start[t]) for t, src in zip(start, codes)} for name, codes in table}
pd.DataFrame(rows).T
```

**표를 읽는 요령**
- 튜플의 추가·삭제·비우기는 모두 ❌ — 튜플은 **바꿀 수 없기** 때문입니다. `x + (4, 5)`는 튜플을 바꾼 것이 아니라 **새 튜플**을 만든 것입니다.
- 집합의 `x[0]`, `x.index()`는 ❌ — 집합은 **순서가 없어** 번호로 꺼낼 수 없습니다.
- 딕셔너리는 번호가 아니라 **키**로 꺼냅니다. 반복하면 기본적으로 **키**가 나오고, 키와 값을 함께 보려면 `x.items()`.
- `remove()`는 없는 값을 지우면 오류가 나지만, 집합의 `discard()`는 오류를 내지 않습니다.

## 5. 실습 C — 항목 추가·수정·삭제 비교
**목표:** 자료형마다 바꾸는 방법이 다르다는 점을 직접 확인합니다.

```python
# 리스트: 추가, 수정, 삭제
a = ["김민수", "이영희"]
a.append("박철수")
a[0] = "최유진"
a.remove("이영희")
print("리스트:", a)

# 튜플: 항목 변경이 불가능하므로 '새 튜플'을 만들어 다시 저장
c = ("김민수", "이영희")
c = c + ("박철수",)
print("새 튜플:", c)

# 집합: 추가와 삭제
d = {"김민수", "이영희"}
d.add("박철수")
d.discard("이영희")
print("집합:", d)

# 딕셔너리: 추가, 값 수정, 키 삭제
b = {"김민수": "출석", "이영희": "출석"}
b["박철수"] = "지각"
b["김민수"] = "지각"
del b["이영희"]
print("딕셔너리:", b)
```

**오류 확인** — 튜플을 직접 바꾸거나, 없는 값을 지우면 어떻게 될까요? 실행해서 메시지를 확인하세요.

```python
try:
    c[0] = "최유진"
except TypeError as e:
    print("❗ 튜플 변경 →", type(e).__name__, ":", e)

try:
    a.remove("없는이름")
except ValueError as e:
    print("❗ 리스트 remove(없는 값) →", type(e).__name__, ":", e)

d.discard("없는이름")
print("✔ 집합 discard(없는 값) → 오류 없음:", d)
```

> 튜플은 기존 항목을 바꾸거나 추가한 것이 아니라, 기존 튜플을 이용해 **새 튜플을 만든 것**입니다. 집합 출력 순서는 달라질 수 있습니다.
