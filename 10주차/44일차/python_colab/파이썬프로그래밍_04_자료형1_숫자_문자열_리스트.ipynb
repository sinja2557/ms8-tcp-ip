## 1. 데이터 타입(자료형)이란?
데이터를 저장하는 **방식**입니다. 짐의 종류에 따라 승용차·버스·기차를 고르듯,
데이터의 성격에 맞는 자료형을 사용합니다. 파이썬은 값을 보고 자료형을 **자동으로** 정합니다.

| 자료형 | 예 |
|---|---|
| 숫자형 | `123`, `3.14` |
| 문자열 | `"hello world"` |
| 리스트 | `[1, 3, 5, 7, 9]` |
| 튜플 | `(1, 2, 3)` |
| 딕셔너리 | `{"name": "Kim", "phone": "010-2222-3333"}` |
| 집합 | `set([1, 2, 3])` |
| 불 | `True`, `False` |

```python
samples = [123, 3.14, "hello", [1, 3, 5], (1, 2), {"name": "Kim"}, {1, 2, 3}, True]
for s in samples:
    print(repr(s), "→", type(s).__name__)
```

## 2. 숫자형
| 종류 | 예 |
|---|---|
| 정수(int) | `123`, `-345`, `0` |
| 실수(float) | `123.45`, `-1234.5`, `3.4e10` (=3.4×10¹⁰) |
| 8진수 | `0o34` (숫자 0 + 알파벳 o) |
| 16진수 | `0x2A` (숫자 0 + 알파벳 x) |

```python
a = -178
b = 4.24e10
c = 4.24e-10
d = 0o177       # 8진수
e = 0x8ff       # 16진수
print(a, b, c)
print("0o177 =", d, "/ 0x8ff =", e)
```

### 숫자 연산
사칙연산 외에 `**`(제곱), `//`(몫), `%`(나머지)를 자주 사용합니다.

```python
a, b = 3, 4
print(a + b, a * b, a / b)
print("3 ** 4 =", a ** b)
print("7 // 4 =", 7 // 4)
print("7 % 3 =", 7 % 3, "/ 3 % 7 =", 3 % 7)
```

## 3. 문자열(string)
문자들의 모임입니다. 4가지 방법으로 만들 수 있습니다.
`"큰따옴표"`, `'작은따옴표'`, `"""큰따옴표 3개"""`, `'''작은따옴표 3개'''`

따옴표 3개를 쓰거나 `\n`(줄바꿈 이스케이프 코드)을 넣으면 여러 줄 문자열을 만들 수 있습니다.

```python
s1 = "Hello World"
s2 = 'Python is fun'
multiline1 = "Life is too short\nYou need python"
multiline2 = """Life is too short
You need python"""
print(s1); print(s2)
print(multiline1)
print(multiline2)
```

### 문자열 연산: 더하기, 곱하기, 길이

```python
head = "python"
tail = " is fun!"
print(head + tail)          # 연결
print(head * 2)             # 반복
print("=" * 30)
a = "Life is too short"
print("길이:", len(a))
```

### 인덱싱과 슬라이싱
- **인덱싱** `a[번호]`: 한 글자를 꺼냄. 번호는 **0부터** 시작, 음수는 뒤에서부터
- **슬라이싱** `a[시작:끝]`: 시작부터 **끝 바로 앞까지** 잘라냄

```python
a = "Life is too short, You need Python"
print(a[0], a[12], a[-1])
print(a[0:4])
print(a[19:])       # 끝 번호 생략 → 끝까지
print(a[:4])        # 시작 번호 생략 → 처음부터
```

```python
# 활용: 날짜와 날씨 분리하기
a = "20010331Rainy"
date = a[:8]
weather = a[8:]
print("날짜:", date, "/ 날씨:", weather)
```

### 문자열 포매팅
문자열 안에 값을 끼워 넣는 방법입니다.

| 코드 | 의미 |
|---|---|
| `%s` | 문자열 |
| `%c` | 문자 1개 |
| `%d` | 정수 |
| `%f` | 실수 |
| `%o` / `%x` | 8진수 / 16진수 |
| `%%` | `%` 문자 자체 |

```python
print("I eat %d apples." % 3)
print("I eat %s apples." % "five")
num, day = 10, "three"
print("I ate %d apples. so I was sick for %s days." % (num, day))
print("rate is %f" % 3.21)
print("변환율 %d%%" % 98)
```

#### 정렬과 소수점 자릿수
- `%10s`: 10칸 확보 후 **오른쪽** 정렬, `%-10s`: **왼쪽** 정렬
- `%0.4f`: 소수점 4자리까지, `%10.4f`: 10칸 확보 + 소수점 4자리

```python
print("[%10s]" % "hi")
print("[%-10s]jane" % "hi")
print("[%0.4f]" % 3.42134234)
print("[%10.4f]" % 3.42134234)
```

#### f-문자열 (파이썬 3.6 이상, 가장 많이 사용)
문자열 앞에 `f`를 붙이고 `{변수}` 또는 `{식}`을 넣습니다.

```python
name = "홍길동"
age = 30
print(f"나의 이름은 {name}입니다. 나이는 {age}입니다.")
print(f"나는 내년이면 {age + 1}살이 된다.")
print(f"원주율은 약 {3.141592:.2f}")
```

### 문자열 관련 함수
| 함수 | 기능 |
|---|---|
| `count(x)` | x의 개수 |
| `find(x)` | x가 처음 나온 위치 (없으면 -1) |
| `join(s)` | 문자 사이에 삽입 |
| `upper()` / `lower()` | 대문자 / 소문자로 |
| `replace(a, b)` | a를 b로 바꿈 |
| `split(sep)` | 구분자로 나누어 리스트로 |

```python
a = "hobby"
print(a.count("b"))
b = "Python is best choice"
print(b.find("b"), b.find("k"))
print(",".join("abcd"))
print("hi".upper(), "HI".lower())
c = "Life is too short"
print(c.replace("Life", "Your leg"))
print(c.split())
print("a:b:c:d".split(":"))
```

## 4. 리스트(list)
여러 값을 **순서대로** 담는 자료형입니다. 대괄호 `[ ]`로 감싸고 쉼표로 구분하며,
어떤 자료형이든(리스트 포함) 담을 수 있습니다. 값의 추가·삭제·검색이 쉽습니다.

```python
a = []
b = [1, 2, 3, 4]
c = ["Life", "is", "too", "short"]
d = [1, 2, "Life", "is"]
e = [1, 2, ["Life", "is"]]
print(a, b, c, d, e, sep="\n")
```

### 인덱싱·슬라이싱·연산
문자열과 같은 방법으로 인덱싱과 슬라이싱을 쓸 수 있고, `+`로 합치고 `*`로 반복합니다.

```python
a = [1, 2, 3]
print(a[0], a[-1], a[0] + a[2])
e = [1, 2, ["Life", "is"]]
print(e[2][0])          # 리스트 안 리스트
b = [4, 5, 6]
print(a + b)
print(a * 3)
print(len(a + b))
```

### 값 수정과 삭제

```python
a = [1, 2, 3]
a[2] = 4
print("수정:", a)

a = [1, 2, 3]
del a[1]
print("del a[1]:", a)

a = [1, 2, 3, 4, 5]
del a[2:]
print("del a[2:]:", a)
```

### 리스트 관련 함수
| 함수 | 기능 |
|---|---|
| `append(x)` | 맨 뒤에 추가 |
| `insert(i, x)` | i번째 위치에 삽입 |
| `extend(list)` | 리스트를 이어 붙임 |
| `remove(x)` | 처음 나오는 x 삭제 |
| `pop()` | 맨 마지막 요소를 꺼내고 삭제 |
| `index(x)` | x의 위치 (없으면 오류) |
| `count(x)` | x의 개수 |
| `sort()` / `reverse()` | 정렬 / 뒤집기 |

```python
a = [1, 2, 3]
a.append(4);        print("append :", a)
a.insert(0, 0);     print("insert :", a)
a.extend([5, 6]);   print("extend :", a)
a.remove(3);        print("remove :", a)
print("pop    :", a.pop(), "→", a)
print("index(4):", a.index(4))
```

```python
a = [1, 4, 3, 2, 1]
print("count(1):", a.count(1))
a.sort();    print("sort   :", a)
a.reverse(); print("reverse:", a)
names = ["a", "c", "b"]
names.sort(); print(names)
```
