## 2. 변수와 기본 자료형
| 자료형 | 의미 | 예 |
|---|---|---|
| `int` | 정수 | `10` |

| `bool` | 참/거짓 | `True`, `False` |
| `None` | 값 없음 | `None` |

- `type(x)` : 자료형 확인
- `int()`, `float()`, `str()`, `bool()` : 형 변환

```python
a = 10; b = 3.14; c = "파이썬"; d = True; e = None
print(type(a), type(b), type(c), type(d), type(e))

print(int("123") + 1)      # 문자열 → 정수 : 124
print(float("2.5") * 2)    # 문자열 → 실수 : 5.0
print(str(100) + "점")     # 정수 → 문자열 : 100점
print(int(3.99))           # 소수점 버림 : 3
```

## 3. 문자열(String) 다루기
- **인덱싱** `s[0]`, **슬라이싱** `s[시작:끝:간격]` (끝은 포함 안 됨)
- **f-string** : `f"{변수}"` 로 값 끼워 넣기
- 주요 메서드 : `upper`, `lower`, `strip`, `replace`, `split`, `join`, `find`, `count`, `startswith`

```python
s = "  Hello Python  "
t = s.strip()                      # 앞뒤 공백 제거
print(t[0], t[-1])                 # H n
print(t[0:5], t[::-1])             # Hello / 거꾸로
print(t.upper(), t.lower())
print(t.replace("Python", "World"))
print(t.find("Py"), t.count("o"), t.startswith("He"))

words = "사과,배,포도".split(",")    # 문자열 → 리스트
print(words)
print(" / ".join(words))           # 리스트 → 문자열

name, age = "페리", 30
print(f"{name}님은 {age}살, 내년엔 {age+1}살")
print(f"원주율 {3.14159:.2f}, 금액 {1234567:,}원")
print(len(t), "글자")
```

## 4. 리스트 (list) `[ ]`
- 순서가 있고, **수정 가능**, 중복 허용
- 주요 기능 : `append`, `insert`, `remove`, `pop`, `sort`, `reverse`, `index`, `in`

```python
fruits = ["사과", "배"]
fruits.append("포도")        # 끝에 추가
fruits.insert(0, "딸기")     # 0번 위치에 추가
print(fruits)

fruits.remove("배")          # 값으로 삭제
last = fruits.pop()          # 마지막 꺼내기
print(fruits, last)

nums = [5, 2, 9, 1]
nums.sort()                  # 오름차순 정렬
print(nums, nums[1:3], len(nums))
print(9 in nums)             # 포함 여부
```

## 5. 튜플 (tuple) `( )`
- 리스트와 비슷하지만 **수정 불가** → 바뀌면 안 되는 데이터에 사용
- 여러 값을 한 번에 나눠 담기(언패킹)에 자주 사용

```python
point = (3, 5)
x, y = point                 # 언패킹
print(x, y, point[0])

a, b = 1, 2
a, b = b, a                  # 값 교환
print(a, b)
# point[0] = 10  → TypeError (수정 불가)
```

## 6. 딕셔너리 (dict) `{키: 값}`
- **키로 값을 찾는** 자료형
- 주요 기능 : `get`, `keys`, `values`, `items`, `update`, `pop`

```python
student = {"name": "홍길동", "age": 20}
student["major"] = "보안"            # 추가
student["age"] = 21                  # 수정
print(student["name"], student.get("phone", "없음"))

for k, v in student.items():
    print(k, ":", v)

print(list(student.keys()), "age" in student)
```

## 10. if 조건문과 for 반복문
- **if / elif / else** : 조건에 따라 분기 (들여쓰기 4칸이 블록)
- **for** : 리스트·문자열·range 등의 요소를 하나씩 꺼내 반복

```python
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"
print("학점:", grade)

# 한 줄 조건식(삼항 연산)
print("합격" if score >= 60 else "불합격")

for fruit in ["사과", "배", "포도"]:
    print(fruit)

for ch in "ABC":
    print(ch, end=" ")
```

## 11. while 반복문 · break · continue
- **while** : 조건이 참인 동안 계속 반복
- **break** : 반복 즉시 종료 / **continue** : 이번 회차만 건너뛰기

```python
n = 1
while n <= 5:
    print(n, end=" ")
    n += 1
print()

for i in range(1, 11):
    if i % 2 == 0:
        continue      # 짝수 건너뛰기
    if i > 7:
        break         # 7 넘으면 종료
    print(i, end=" ")
```

## 12. range · enumerate · zip
- `range(시작, 끝, 간격)` : 숫자 범위 생성 (끝은 포함 안 됨)
- `enumerate()` : **번호(인덱스)와 값**을 함께 꺼냄
- `zip()` : 여러 리스트를 **짝지어** 꺼냄

```python
print(list(range(5)))          # [0, 1, 2, 3, 4]
print(list(range(1, 10, 2)))   # [1, 3, 5, 7, 9]
print(list(range(10, 0, -3)))  # [10, 7, 4, 1]

for i, name in enumerate(["철수", "영희"], start=1):
    print(i, name)

names = ["철수", "영희"]; scores = [90, 85]
for n, s in zip(names, scores):
    print(n, s)
```

## 13. 컴프리헨션 (Comprehension)
반복문 + 조건을 **한 줄로** 리스트/딕셔너리 만들기

```python
squares = [x**2 for x in range(1, 6)]
evens = [x for x in range(10) if x % 2 == 0]
word_len = {w: len(w) for w in ["sky", "python"]}
print(squares, evens, word_len)
```

## 14. 함수 정의 `def` · `lambda`
- `def 함수명(매개변수):` 로 정의, `return` 으로 결과 반환
- 기본값 매개변수, 가변 인자 `*args`, 키워드 인자 `**kwargs`
- `lambda` : 이름 없는 한 줄 함수

```python
def add(a, b=0):
    """두 수를 더해 반환"""
    return a + b

print(add(3, 5), add(3))

def total(*args):
    return sum(args)
print(total(1, 2, 3, 4))

def profile(**kwargs):
    for k, v in kwargs.items():
        print(k, "=", v)
profile(name="홍길동", city="서울")

def min_max(nums):
    return min(nums), max(nums)   # 여러 값 반환(튜플)
lo, hi = min_max([4, 1, 9])
print(lo, hi)

square = lambda x: x * x
print(square(7))
```

## 15. 자주 쓰는 내장 함수
| 함수 | 기능 |
|---|---|
| `len` | 길이 |
| `sum / max / min` | 합계 / 최댓값 / 최솟값 |
| `sorted` | 정렬된 새 리스트 |
| `abs / round` | 절댓값 / 반올림 |

```python
nums = [3, -1, 7, 2]
print(len(nums), sum(nums), max(nums), min(nums))
print(sorted(nums), sorted(nums, reverse=True))
print(abs(-5), round(3.14159, 2))
print(list(map(str, nums)))
print(list(filter(lambda x: x > 0, nums)))
print(isinstance(3.0, float))
```

## 16. 예외 처리 `try / except`
오류가 나도 프로그램이 멈추지 않도록 처리
- `except 오류종류` : 해당 오류 처리, `else` : 오류 없을 때, `finally` : 항상 실행

```python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("0으로 나눌 수 없습니다")
        return None
    except TypeError as e:
        print("자료형 오류:", e)
        return None
    else:
        return result
    finally:
        print("계산 종료")

print(divide(10, 2))
print(divide(10, 0))
```

## 17. 클래스 `class`
- 데이터(속성)와 기능(메서드)을 묶은 **설계도**, 이 설계도로 만든 것이 **객체(인스턴스)**
- `__init__` : 객체 생성 시 자동 실행(초기화), `self` : 자기 자신
- **상속** : 기존 클래스를 물려받아 확장

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def __str__(self):
        return f"{self.owner} 계좌 잔액: {self.balance:,}원"

acc = Account("홍길동", 1000)
acc.deposit(5000)
print(acc)

class SavingAccount(Account):           # 상속
    def __init__(self, owner, balance=0, rate=0.03):
        super().__init__(owner, balance)
        self.rate = rate

    def add_interest(self):
        self.balance += int(self.balance * self.rate)

s = SavingAccount("김철수", 10000)
s.add_interest()
print(s)
```

## 18. 파일 읽기 · 쓰기
- `open(파일명, 모드, encoding="utf-8")`
  - `"w"` 쓰기(덮어쓰기), `"a"` 이어쓰기, `"r"` 읽기
- **`with` 문** 사용 → 작업이 끝나면 파일이 자동으로 닫힘
- 읽기 : `read()` 전체, `readline()` 한 줄, `readlines()` 줄 목록, 또는 `for line in f`

```python
# 쓰기
with open("memo.txt", "w", encoding="utf-8") as f:
    f.write("첫째 줄\n")
    f.write("둘째 줄\n")

# 이어쓰기
with open("memo.txt", "a", encoding="utf-8") as f:
    f.write("셋째 줄\n")

# 전체 읽기
with open("memo.txt", "r", encoding="utf-8") as f:
    print(f.read())

# 한 줄씩 읽기
with open("memo.txt", encoding="utf-8") as f:
    for i, line in enumerate(f, 1):
        print(i, line.strip())
```

### 18-1. CSV · JSON 파일
- **CSV** : 표 형태 데이터 → `csv` 모듈
- **JSON** : 딕셔너리/리스트 구조 데이터 → `json` 모듈 (`ensure_ascii=False` 로 한글 그대로 저장)
- 파일 존재 확인 : `os.path.exists()`

```python
import csv, json, os

rows = [["이름", "점수"], ["철수", 90], ["영희", 85]]
with open("scores.csv", "w", newline="", encoding="utf-8") as f:
    csv.writer(f).writerows(rows)

with open("scores.csv", encoding="utf-8") as f:
    for row in csv.reader(f):
        print(row)

data = {"name": "홍길동", "skills": ["Python", "보안"]}
with open("data.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

with open("data.json", encoding="utf-8") as f:
    loaded = json.load(f)
print(loaded["skills"])

print(os.path.exists("data.json"), os.listdir(".")[:5])
```

## 20. Enum (열거형)
정해진 값들의 목록에 **이름을 붙여** 관리 (예: 상태, 요일, 등급)

```python
from enum import Enum

class Status(Enum):
    READY = 1
    RUNNING = 2
    DONE = 3

s = Status.RUNNING
print(s, s.name, s.value)
print(s == Status.RUNNING)
for st in Status:
    print(st.name, st.value)
```

## ✅ 종합 실습 — 성적 관리 미니 프로그램
지금까지 배운 dict · for · if · 함수 · 파일을 함께 사용합니다.

```python
import json

students = {"철수": [90, 85, 77], "영희": [95, 92, 88], "민수": [60, 72, 55]}

def average(scores):
    return sum(scores) / len(scores)

def grade(avg):
    if avg >= 90: return "A"
    elif avg >= 80: return "B"
    elif avg >= 70: return "C"
    return "F"

report = {}
for name, scores in students.items():
    avg = average(scores)
    report[name] = {"평균": round(avg, 1), "학점": grade(avg)}
    print(f"{name:4s} 평균 {avg:5.1f}  학점 {grade(avg)}")

with open("report.json", "w", encoding="utf-8") as f:
    json.dump(report, f, ensure_ascii=False, indent=2)
print("report.json 저장 완료")
```
