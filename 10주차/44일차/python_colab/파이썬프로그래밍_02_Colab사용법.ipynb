## 1. 클라우드 개발 환경, Colab
Google Colab은 웹 브라우저만 있으면 설치 없이 파이썬을 실행할 수 있는 **클라우드 개발 환경**입니다.
노트북은 두 종류의 셀로 이루어집니다.

- **텍스트 셀**: 지금 읽고 있는 설명 셀 (마크다운 문법)
- **코드 셀**: 파이썬 코드를 입력하고 실행하는 셀

| 단축키 | 기능 |
|---|---|
| `Shift + Enter` | 현재 셀 실행 후 다음 셀로 이동 |
| `Ctrl + Enter` | 현재 셀 실행 (이동 없음) |
| `Ctrl + M, B` | 아래에 코드 셀 추가 |
| `Ctrl + /` | 선택한 줄 주석 처리 |

```python
print("첫 번째 코드 셀입니다.")
```

## 2. 셀의 마지막 줄은 자동으로 출력
코드 셀의 **마지막 줄이 값**이면 `print()` 없이도 결과가 표시됩니다.

```python
10 + 20
```

## 3. 셀 사이에서 변수 공유
한 셀에서 만든 변수는 다른 셀에서도 사용할 수 있습니다. 단, **먼저 실행한 셀**의 결과가 기억되므로
위에서부터 순서대로 실행하는 습관을 들이세요.

```python
message = "Colab에서 만든 변수"
```

```python
print(message)
```

## 4. 주석
`#` 뒤의 내용은 실행되지 않는 **주석**입니다. 코드에 설명을 달 때 사용합니다.

```python
# 이 줄은 실행되지 않습니다.
print("주석 아래 줄만 실행됩니다.")   # 줄 끝에도 주석을 달 수 있습니다.
```

## 5. 셸 명령어
