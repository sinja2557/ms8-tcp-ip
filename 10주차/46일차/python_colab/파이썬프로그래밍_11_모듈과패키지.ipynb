## 1. 모듈이란?
함수·변수·클래스를 모아 놓은 **파이썬 파일(.py)** 입니다. 다른 프로그램에서 `import`로 불러와 사용합니다.

### 모듈 만들기
Colab에서는 `%%writefile 파일이름` 을 셀 맨 위에 쓰면 셀 내용이 파일로 저장됩니다.
(Colab 기본 폴더는 `/content` 입니다.)

````python
%%writefile mod1.py
def add(a, b):
    return a + b

def sub(a, b):
    return a - b
````

## 2. 모듈 불러오기
| 방법 | 사용 |
|---|---|
| `import mod1` | `mod1.add(3, 4)` |
| `from mod1 import add` | `add(3, 4)` (모듈 이름 없이) |
| `from mod1 import add, sub` | 여러 개 |
| `from mod1 import *` | 모두 (`*` = 모든 것) |

모듈 이름은 파일 이름에서 `.py`를 뺀 것입니다.

````python
import mod1
print(mod1.add(3, 4))
print(mod1.sub(4, 2))

from mod1 import add, sub
print(add(10, 20), sub(10, 20))
````

## 3. `if __name__ == "__main__":`
`__name__`은 파이썬이 내부적으로 쓰는 특별한 변수입니다.
- 파일을 **직접 실행**하면 `__name__` 값은 `"__main__"` → 아래 블록이 실행됨
- 다른 파일에서 **import** 하면 `__name__` 값은 모듈 이름(`"mod1"`) → 실행되지 않음

테스트 코드를 모듈에 넣어 둘 때 사용합니다.

````python
%%writefile mod1.py
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

if __name__ == "__main__":
    print("직접 실행:", add(1, 4), sub(4, 2))
````

````python
# 직접 실행 → 출력 O
!python mod1.py
````

````python
# import → 출력 X (이미 불러온 모듈을 다시 읽기 위해 reload 사용)
import importlib, mod1
importlib.reload(mod1)
print("__name__ 값:", mod1.__name__)
````

## 4. 클래스와 변수가 포함된 모듈
원의 넓이를 구하는 `Math` 클래스, `add` 함수, 원주율 `PI` 변수를 담은 `mod2.py`를 만들어 봅시다.

````python
%%writefile mod2.py
PI = 3.141592

class Math:
    def solv(self, r):
        return PI * (r ** 2)

def add(a, b):
    return a + b
````

````python
import mod2
print(mod2.PI)
a = mod2.Math()
print(a.solv(2))
print(mod2.add(mod2.PI, 4.4))
````

## 5. 다른 폴더에 있는 모듈 불러오기 – `sys.path`
파이썬은 `sys.path`에 등록된 폴더에서 모듈을 찾습니다. 다른 폴더의 모듈을 쓰려면 경로를 추가합니다.

````python
import os, sys
os.makedirs("sample", exist_ok=True)
with open("sample/mod3.py", "w", encoding="utf-8") as f:
    f.write("def hello():\n    return 'sample 폴더의 mod3 모듈입니다.'\n")

sys.path.append(os.path.abspath("sample"))
print(sys.path[-1])

import mod3
print(mod3.hello())
````

## 6. 패키지란?
여러 모듈을 **폴더(디렉터리) 구조**로 묶은 것입니다. `패키지.모듈` 처럼 점(`.`)으로 계층을 표현합니다.

```
game/
├── __init__.py
├── sound/
│   ├── __init__.py
│   └── echo.py
└── graphic/
    ├── __init__.py
    └── render.py
```
- 큰 프로그램을 나누어 관리하기 쉽고 공동 작업·유지 보수에 유리
- 다른 모듈과 이름이 겹쳐도 안전
- 각 폴더의 `__init__.py`는 해당 폴더가 패키지임을 알려 주는 파일

### 패키지 만들기

````python
import os
for d in ["game", "game/sound", "game/graphic"]:
    os.makedirs(d, exist_ok=True)
    open(f"{d}/__init__.py", "w").close()
print(os.listdir("game"))
````

````python
%%writefile game/sound/echo.py
def echo_test():
    print("echo")
````

````python
%%writefile game/graphic/render.py
def render_test():
    print("render")
````

### 패키지 안의 함수 실행하기 – 3가지 방법

````python
import sys, os
sys.path.append(os.getcwd())      # 현재 폴더(/content)를 검색 경로에 추가

# ① 모듈을 import
import game.sound.echo
game.sound.echo.echo_test()

# ② 모듈이 있는 패키지에서 from ... import
from game.sound import echo
echo.echo_test()

# ③ 함수를 직접 import
from game.sound.echo import echo_test
echo_test()

from game.graphic.render import render_test
render_test()
````
