## 1. 클래스와 객체
- **클래스(class)**: 변수와 함수를 묶어 정의한 **설계도·템플릿** (예: 문서 양식)
- **객체(object)**: 클래스로 만들어 낸 실체 (예: 양식으로 작성한 A 보고서, B 보고서)

같은 클래스로 만든 객체라도 **각자 고유한 값**을 가지며 서로 영향을 주지 않습니다.

````python
class Cookie:
    pass            # 아무 내용 없는 클래스

a = Cookie()
b = Cookie()
print(type(a))
print(a is b)       # 서로 다른 객체
````

## 2. 클래스의 구성 요소
| 요소 | 설명 |
|---|---|
| 메서드 | 클래스 안에 정의된 함수 |
| 클래스 변수 | 클래스 안에 정의된 변수 (모든 객체가 공유) |
| 인스턴스(객체) 변수 | `self.이름`으로 객체마다 따로 만든 변수 |

````python
class Stock:
    market = "kospi"                  # 클래스 변수

    def transaction(self, name):      # 메서드
        self.name = name              # 인스턴스 변수
        print(f"{self.name} 주식을 매수했습니다.")

s = Stock()
s.transaction("삼성전자")
print(s.market, s.name)
````

### 클래스 변수와 인스턴스 변수의 차이
객체에서 이름을 찾을 때 **먼저 객체 자신**을 보고, 없으면 **클래스**에서 찾습니다.

````python
class Stock:
    market = "kospi"
    def check_market(self):
        print(self.market)

s1 = Stock()
s2 = Stock()
s1.market = "kosdaq"      # s1에만 인스턴스 변수 생성
s1.check_market()         # kosdaq
s2.check_market()         # kospi (클래스 변수)
````

## 3. 사칙연산 클래스 FourCal 만들기
목표: 숫자 2개를 지정하고 더하기·빼기·곱하기·나누기를 할 수 있는 클래스

```python
a = FourCal()
a.setdata(4, 2)
a.add()   # 6
```
### 단계 ① 클래스 구조 만들기

````python
class FourCal:
    pass

a = FourCal()
print(type(a))
````

### 단계 ② 숫자를 지정하는 setdata 메서드
메서드의 첫 번째 매개변수 **`self`에는 메서드를 호출한 객체가 자동으로 전달**됩니다.
그래서 `a.setdata(4, 2)`처럼 2개만 넘겨도 `self=a, first=4, second=2`가 됩니다.

결과적으로 `a.first = 4`, `a.second = 2` 라는 객체 변수가 생깁니다.

````python
class FourCal:
    def setdata(self, first, second):
        self.first = first
        self.second = second

a = FourCal()
a.setdata(4, 2)
print(a.first, a.second)
````

### 객체 변수는 객체마다 따로 저장된다
`b.first`를 3으로 바꾸면 `a.first`도 바뀔까요?

````python
a = FourCal()
b = FourCal()
a.setdata(4, 2)
b.setdata(3, 7)
print("a.first =", a.first)    # 4 그대로
print("b.first =", b.first)
````

### 단계 ③ 더하기, ④ 빼기·곱하기·나누기 메서드

````python
class FourCal:
    def setdata(self, first, second):
        self.first = first
        self.second = second
    def add(self):
        return self.first + self.second
    def sub(self):
        return self.first - self.second
    def mul(self):
        return self.first * self.second
    def div(self):
        return self.first / self.second

a = FourCal()
a.setdata(4, 2)
print(a.add(), a.sub(), a.mul(), a.div())

b = FourCal()
b.setdata(3, 8)
print(b.add(), b.sub(), b.mul(), b.div())
````

## 4. 생성자 `__init__` – 객체를 만들 때 바로 값 지정
`setdata`를 따로 부르지 않아도, 객체를 만드는 순간 실행되는 특별한 메서드입니다.

````python
class FourCal:
    def __init__(self, first, second):
        self.first = first
        self.second = second
    def add(self):
        return self.first + self.second
    def div(self):
        if self.second == 0:
            return "0으로 나눌 수 없습니다."
        return self.first / self.second

a = FourCal(10, 0)
print(a.add(), a.div())
````

## 5. 실습: 은행 계좌 클래스
입금·출금·잔액 조회 기능을 가진 클래스를 만들어 봅시다.

````python
class Account:
    bank = "파이썬은행"                 # 클래스 변수

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, money):
        self.balance += money
        print(f"{money:,}원 입금 → 잔액 {self.balance:,}원")

    def withdraw(self, money):
        if money > self.balance:
            print("잔액이 부족합니다.")
        else:
            self.balance -= money
            print(f"{money:,}원 출금 → 잔액 {self.balance:,}원")

acc = Account("홍길동", 10000)
acc.deposit(5000)
acc.withdraw(20000)
acc.withdraw(3000)
print(acc.bank, acc.owner, acc.balance)
````
