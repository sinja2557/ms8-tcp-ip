## 1. 반복문이 필요한 이유
학생이 100명이라면 같은 성적 계산을 100번 써야 할까요? **반복문**을 사용하면 한 번만 작성하여 여러 번 실행할 수 있습니다.

## 2. while 문
```python
while 조건:
    조건이 참인 동안 반복할 문장들
```
"열 번 찍어 안 넘어가는 나무 없다" – 나무를 찍은 횟수가 10보다 작은 동안 반복합니다.

````python
treeHit = 0
while treeHit < 10:
    treeHit = treeHit + 1
    print(f"나무를 {treeHit}번 찍었습니다.")
    if treeHit == 10:
        print("나무 넘어갑니다.")
````

### break – 반복문 강제로 빠져나가기
자판기에 커피가 10잔 있고 300원을 넣을 때마다 한 잔씩 줄어듭니다.
`money`가 항상 300이므로 조건은 늘 참(무한 루프) → 커피가 떨어지면 `break`로 종료합니다.

````python
coffee = 10
money = 300
while money:
    print("돈을 받았으니 커피를 줍니다.")
    coffee = coffee - 1
    print(f"남은 커피의 양은 {coffee}개입니다.")
    if coffee == 0:
        print("커피가 다 떨어졌습니다. 판매를 중지합니다.")
        break
````

### continue – 반복문의 처음으로 돌아가기
1부터 10까지 중 **홀수만** 출력합니다. 짝수이면 `continue`로 아래 `print`를 건너뜁니다.

````python
a = 0
while a < 10:
    a = a + 1
    if a % 2 == 0:
        continue
    print(a)
````

## 3. for 문
```python
for 변수 in 리스트(또는 튜플, 문자열):
    반복할 문장들
```
리스트의 첫 요소부터 마지막 요소까지 차례로 변수에 대입되며 블록이 실행됩니다.

````python
test_list = ["one", "two", "three"]
for i in test_list:
    print(i)
````

````python
# 튜플 언패킹과 함께 쓰기
pairs = [(1, 2), (3, 4), (5, 6)]
for first, last in pairs:
    print(first + last)

for ch in "Python":
    print(ch, end=" ")
````

### range 함수
| 형태 | 결과 |
|---|---|
| `range(10)` | 0, 1, …, 9 |
| `range(1, 11)` | 1, 2, …, 10 (끝 숫자는 포함 X) |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 (간격 지정) |

````python
print(list(range(10)))
print(list(range(1, 11)))
print(list(range(0, 10, 2)))

total = 0
for i in range(1, 11):
    total += i
print("1~10의 합:", total)
````

### 실습: 구구단
바깥 `for`는 단(2~9), 안쪽 `for`는 곱하는 수(1~9)를 담당합니다.

````python
for i in range(2, 10):
    for j in range(1, 10):
        print(f"{i*j:3d}", end=" ")
    print()
````

### 리스트 내포 (List Comprehension)
`[표현식 for 항목 in 반복가능객체 if 조건]` – 반복문으로 리스트를 만드는 코드를 한 줄로 줄여 줍니다.

````python
a = [1, 2, 3, 4]

# 일반 for 문
result = []
for num in a:
    result.append(num * 3)
print(result)

# 리스트 내포
print([num * 3 for num in a])

# if 조건 – 짝수에만 3을 곱함
print([num * 3 for num in a if num % 2 == 0])

# for 2개 – 구구단 결과를 모두 리스트에
gugu = [x * y for x in range(2, 10) for y in range(1, 10)]
print(len(gugu), gugu[:9])
````

## 4. 실습: 학생 성적 처리
학생 한 명의 성적을 `[퀴즈, 중간고사, 레포트, 기말고사]` 리스트로, 여러 학생을 리스트의 리스트로 표현합니다.

````python
score = [[23, 90, 20, 90], [25, 50, 20, 10], [0, 10, 0, 30], [5, 70, 5, 100]]
grade = []

for i in score:          # 학생 한 명
    total = 0
    for j in i:          # 성적 요소
        total += j
    grade.append(total)

print(grade)
````

````python
# 응용: 합계에 따라 등급 매기기
for n, g in enumerate(grade, start=1):
    if g >= 150:
        level = "A"
    elif g >= 100:
        level = "B"
    else:
        level = "C"
    print(f"{n}번 학생: 합계 {g:3d}점 → {level}")
````
