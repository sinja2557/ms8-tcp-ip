## 1. 파이썬이란?
파이썬(Python)은 1991년 귀도 반 로섬이 발표한 **인터프리터 방식**의 프로그래밍 언어입니다.
작성한 코드를 한 줄씩 해석하여 바로 실행하므로, 컴파일 과정 없이 결과를 즉시 확인할 수 있습니다.

가장 먼저 화면에 글자를 출력해 봅시다. `print()`는 괄호 안의 값을 화면에 보여 주는 함수입니다.

```python
print("Hello, Python!")
print("파이썬 프로그래밍을 시작합니다.")
```

## 2. 파이썬으로 할 수 있는 일
| 분야 | 예 |
|---|---|
| 시스템 유틸리티 | 파일·폴더 정리, 자동화 스크립트 |
| GUI 프로그래밍 | Tkinter 등으로 창 프로그램 제작 |
| C/C++ 연동 | 속도가 필요한 부분은 C로, 나머지는 파이썬으로 |
| 웹 프로그래밍 | Django, Flask 등 웹 서비스 |
| 수치 연산·인공지능 | NumPy, pandas, TensorFlow |
| 데이터베이스 | SQLite, MySQL 등과 연동 |

아래는 파이썬에 기본으로 들어 있는 `math` 모듈로 간단한 수치 계산을 해 보는 예입니다.

```python
import math

print("원주율:", math.pi)
print("2의 제곱근:", math.sqrt(2))
print("반지름 3인 원의 넓이:", math.pi * 3 ** 2)
```

## 3. 파이썬의 특징
### (1) 사람의 말과 비슷한 문법
"4가 [1, 2, 3, 4] 안에 있으면 출력하라"를 거의 그대로 코드로 옮길 수 있습니다.

```python
if 4 in [1, 2, 3, 4]:
    print("4가 있습니다")
```

### (2) 간결함
같은 일을 하는 코드가 다른 언어보다 짧습니다. 예를 들어 두 변수의 값을 바꿀 때 임시 변수가 필요 없습니다.

```python
a, b = 1, 2
a, b = b, a      # 값 교환
print(a, b)
```

### (3) 들여쓰기(indentation)로 블록을 구분
파이썬은 `{ }` 대신 **들여쓰기(보통 공백 4칸)** 로 코드 블록을 구분합니다.
들여쓰기가 맞지 않으면 오류가 나므로, 자연스럽게 읽기 좋은 코드가 됩니다.

```python
for i in range(3):
    print("반복", i)      # for 블록 안 (들여쓰기 O)
print("반복 끝")          # for 블록 밖 (들여쓰기 X)
```

### (4) 무료·오픈소스, 빠른 개발 속도
누구나 무료로 사용할 수 있고, 수많은 라이브러리가 공개되어 있어 원하는 기능을 빠르게 만들 수 있습니다.
Colab에는 자주 쓰는 라이브러리가 이미 설치되어 있습니다.

```python
import sys
print("현재 파이썬 버전:", sys.version)
```
